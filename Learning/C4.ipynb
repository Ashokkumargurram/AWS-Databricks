```python
file_path = "/Volumes/workspace/default/learning/customers_phones.csv"
```

```python
customer = spark.read.csv(file_path , header = True, inferSchema = True)
display(customer.limit(5))
```

```python
from pyspark.sql.functions import udf , col
from pyspark.sql.types import StringType
```

```python
def n_p (phone, country):
    if phone is None:
        return None
    digit = ''.join(c for c in phone if c.isdigit())
    codes = {"US": "1" , "UK":"44" , "ES":"34", "DE":"49", "FR":"33", "JP":"81"
             }
    country_code= codes.get(country, "")
    if not digit.startswith(country_code) and country_code:
        digit = country_code + digit
    return "+" + digit
nphone = udf(n_p, StringType())
cleaned = customer.withColumn("NP", nphone(col("phone"), col("country")))
display(cleaned)
```
