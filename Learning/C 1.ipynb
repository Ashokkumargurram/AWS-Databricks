```python
file_path = "/Volumes/workspace/default/learning/employees.csv"
```

```python
emp = spark.read.csv(file_path, header=True, inferSchema=True)
```

```python
emp.show()
```

```python
emp.printSchema()
```

```python
from pyspark.sql.functions import col, avg , round
average = emp.groupBy(col("department")).agg(avg(col("salary")).alias("averagesal"))
```

```python
average.show()
```

```python
avg = emp.groupBy("department").agg(round(avg("salary"), 2).alias("avgsal"))
```

```python
avg.show(5)
```

```python
%sql
select department, round(avg(salary), 2) as avg
from read_files('/Volumes/workspace/default/learning/employees.csv',
                format => 'csv',
                header => true,
                inferSchema => true
)
group by department
```

How many employees are in each department?

```python
from pyspark.sql.functions import countDistinct
dis = emp.groupBy("department").agg(countDistinct("employee_id"))
dis.show()
```

```python
from pyspark.sql.functions import count
emp.groupBy("department").agg(count("employee_id")).show()
```

```python
%sql
select department, count(employee_id) from read_files('/Volumes/workspace/default/learning/employees.csv',
                format => 'csv',
                header => true,
                inferSchema => true
) 
group by department;
```

What is the total salary cost per department?

```python
from pyspark.sql.functions import sum
emp.groupby("department").agg(sum(emp["salary"].cast("float"))).show()
```

```python
emp.groupBy("department").agg(sum("salary")).show()
```

Which department has the highest-paid employee?

```python
%sql
select department, max(salary) from read_files('/Volumes/workspace/default/learning/employees.csv',
                format => 'csv',
                header => true,
                inferSchema => true
)
group by department
order by max(salary)
```
