```python
file_path = "/Volumes/workspace/default/learning/orders.csv"
```

```python
order = spark.read.csv(file_path, header = True, inferSchema = True)
display(order.limit(5))
```

```python
from pyspark.sql.functions import col, trim, initcap 
clean_name = order.withColumn("customer_name",initcap(trim(col("customer_name"))))
display(clean_name)
```

```python
from pyspark.sql.functions import col, when, lit , round
total_value = clean_name.withColumn("order_total", round(col("quantity") * col("unit_price"), 2))
display(total_value)
```

```python
total_value = clean_name.withColumn("order_total",round(col("quantity")*col("unit_price"), 2)) \
    .withColumn("priority", when(col("quantity")>=5, lit(True)).otherwise(lit(False)))
display(total_value)
```

```python
valid = total_value.filter(col("quantity")> 0)
display(valid)
```
