```python
dbutils.widgets.text("file_path_value", "provide your file path")
file = dbutils.widgets.get("file_path_value")
```

```python
df = (
    spark.read
    .option("header", "true")
    .option("inferSchema", "true")
    .csv(file)
)

row_count = df.count()
print(f"Number of rows: {row_count}")

dbutils.notebook.exit(row_count)
```
